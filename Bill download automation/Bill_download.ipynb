```python
#Import all librariess
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from time import sleep
import pandas as pd
import os
import time
```

```python
df=pd.read_excel("Hines-Engie-339256.xlsx") #Import Excel file with links to be downloaded
```

```python
#Function to wait until the present download is completed
def download_wait(path_to_downloads):
    seconds = 0
    dl_wait = True
    while dl_wait and seconds < 20:
        time.sleep(1)
        dl_wait = False
        for fname in os.listdir(path_to_downloads):
            if fname.endswith('.crdownload'):
                dl_wait = True
        seconds += 1
    return seconds
```

```python
chromeOptions=Options()
#Setting the default pdf action as download
chromeOptions.add_experimental_option("prefs", {"download.default_directory":download_path ,
"plugins.always_open_pdf_externally": True})
```

```python
#Function to download the pdf
def pdf_download(filename,link):
    driver = webdriver.Chrome(options=chromeOptions)
    driver.maximize_window()
    driver.get(link)
    time.sleep(1)
    download_wait(download_path)
    os.rename(download_path+'BillImage.pdf', download_path+filename+'.pdf')#Setting invoice number as filename
    driver.close()
```

```python
for i in range(len(df)):
    try:
        pdf_download(df['Filename'][i],df['BILL IMAGE'][i])
    except:
        driver.close()
        print('Error - iteration {}'.format(i))
        continue
```
