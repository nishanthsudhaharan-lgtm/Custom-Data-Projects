#### Credentials

```python
with open('Credentials/Page1.json') as json_file:
    data = json.load(json_file)
```

```python
Creds=pd.DataFrame(data['_embedded']['credentials'])
```

```python
Creds.drop(Creds.index,inplace=True)
```
