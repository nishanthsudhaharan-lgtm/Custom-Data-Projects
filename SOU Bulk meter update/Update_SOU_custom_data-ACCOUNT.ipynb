```python
import pandas as pd
import requests
import getpass
from datetime import date as dt
```

```python
username=str(input('Type Your Username:\n'))
print()
password=str(getpass.getpass(prompt='Password'))
```

```python
url = "https://api.urjanet.com/auth/login"
payloadl = {
    "username": username,
    "password": password
    }
h = {
        "accept": "application/json",
        "content-type": "application/json"
    }
def Api_tkn_request(payloadl):
    
    url = "https://api.urjanet.com/auth/login"
    resd = requests.post(url, json=payloadl, headers=h)
    #print(resd.json())
    tkn=resd.json()['token']
    headers = {
      'Accept': 'application/json',
      'Authorization': 'Bearer '+tkn}
    return headers
```

```python
headers=Api_tkn_request(payloadl)
```

```python
df=pd.read_excel("Accounts_sou.xlsx")
df
```

```python
headers=Api_tkn_request(payloadl)
dc=[]
for i in range(len(df)):
    
    try:
        url = "https://api.urjanet.com/utility/meters?search=accountNumbers%3D%3D"+df["Account Number"][i]+"&page=0&size=100"
        
        response = requests.get(url, headers=headers)
        #print(response.json())
        
        for j in response.json()['_embedded']['meters']:
            dc.append({"AccountNUmber":df["Account Number"][i],"MeterID":j['entityId'],"espm_property_id":df['espm_property_ids'][i],"espm_meter_id":df['espm_meter_ids'][i]})
    except:
        dc.append({"AccountNUmber":df["Account Number"][i],"MeterID":"Not Found","espm_property_id":df['espm_property_ids'][i],"espm_meter_id":df['espm_meter_ids'][i]})
dc=pd.DataFrame(dc)
print(dc)
```

```python
headers=Api_tkn_request(payloadl)
for i in range(len(dc)):
    if dc['MeterID'][i]!="Not Found":
        url = "https://api.urjanet.com/utility/meters/"+dc["MeterID"][i]
        
        payload = { "customData": {
                "espm_meter_id": int(dc['espm_meter_id'][i]),
                "espm_property_id": int(dc['espm_property_id'][i])
            } }
        
        headers=Api_tkn_request(payloadl)
        response = requests.patch(url, json=payload, headers=headers)
        
        print(response.text)
    
        
```
